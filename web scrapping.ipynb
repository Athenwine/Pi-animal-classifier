```python
from bing_image_downloader import downloader
import os

animal_classes = {
    "Mammiferes": ["Lion", "Tigre", "Éléphant", "Girafe", "Zèbre", "Ours", "Chien", "Chat", "Singe", "Cheval"],
    "Oiseaux": ["Aigle", "Perroquet", "Hibou", "Manchot", "Flamant rose", "Canard"],
    "Reptiles": ["Serpent", "Tortue", "Crocodile", "Lézard"],
    "Amphibiens": ["Grenouille", "Salamandre"],
    "Poissons": ["Requin", "Poisson rouge", "Saumon", "Poisson-clown"],
    "Insectes": ["Papillon", "Abeille", "Fourmi", "Coccinelle"]
}

BASE_DIR = "dataset_bing"
NUM_IMAGES = 50

os.makedirs(BASE_DIR, exist_ok=True)
```

```python
for class_name, animals in animal_classes.items():
    print(f"\n📦 Classe : {class_name}")

    class_folder = os.path.join(BASE_DIR, class_name)
    os.makedirs(class_folder, exist_ok=True)

    for animal in animals:
        print(f"  📥 Téléchargement : {animal}")

        animal_folder = os.path.join(class_folder, animal.replace(" ", "_"))
        
        downloader.download(
            query=animal + " animal",
            limit=NUM_IMAGES,
            output_dir=BASE_DIR,
            adult_filter_off=True,
            force_replace=False,
            timeout=10
        )
```

```python
import os

root = r"C:\Users\Aws\Documents\pi 2mba\dataset_bing"

for folder, subfolders, files in os.walk(root):
    for sub in subfolders:
        if "animal" in sub.lower():
            old_path = os.path.join(folder, sub)
            new_name = sub.replace(" animal", "").replace(" Animal", "")
            new_path = os.path.join(folder, new_name)
            os.rename(old_path, new_path)
            print(f"Renamed: {sub} -> {new_name}")
```

```python
from PIL import Image
import os

root = r"C:\Users\Aws\Documents\pi 2mba\dataset_bing"

bad = 0
for folder, _, files in os.walk(root):
    for f in files:
        path = os.path.join(folder, f)
        try:
            img = Image.open(path)
            img.verify()
        except:
            print("Deleted corrupt:", path)
            os.remove(path)
            bad += 1

print("🔥 Total corrupt removed:", bad)
```
